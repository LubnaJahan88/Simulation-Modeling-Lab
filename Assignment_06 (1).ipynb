```python
#LUBNA JAHAN LIPA
#ID: 0432310005101088


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from scipy.stats import ttest_1samp
```

```python
scores = np.array([93,83,69,62,75,93,73,77,65,65,78,90,94,78,57,76,56,78,84,92])
df = pd.DataFrame({
    "Student_ID": range(1, 21),
    "Exam_Score": scores
})
print(df)
```

```python
mean_score = np.mean(scores)
population_variance_score = np.var(scores)
sample_variance_score = np.var(scores, ddof=1)
std_score = np.std(scores, ddof=1)

print("\nExam Score Statistics")
print("Mean:", mean_score)
print("Population Variance:", population_variance_score)
print("Sample Variance:", sample_variance_score)
print("Standard Deviation:", std_score)
```

```python
np.random.seed(42)
generated_scores = np.random.normal(
    loc=mean_score,
    scale=std_score,
    size=20
)
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.hist(scores, bins=6, color='skyblue', edgecolor='black')
plt.title("Real Exam Scores")
plt.subplot(1,2,2)
plt.hist(generated_scores, bins=6, color='orange', edgecolor='black')
plt.title("Generated Exam Scores")
plt.tight_layout()
plt.show()
```

```python
simulated_scores = np.random.normal(
    loc=np.mean(scores),
    scale=np.std(scores, ddof=1),
    size=100000
)
probability = np.mean(simulated_scores >= 75)
print("Probability(score >= 75):", probability)
```

```python
print("Monte Carlo Simulation")
print("Probability(score >= 75):", simulated_scores)
```

```python
student_id_last3 = 88
hypothesized_mean_score = (student_id_last3 % 50) + 50
```

```python
t_stat_score, p_value_score = ttest_1samp(scores, hypothesized_mean_score)
```

```python
print("\nHypothesis Testing for Exam Scores")
print("Hypothesized Mean:", hypothesized_mean_score)
print("T-Statistic:", t_stat_score)
print("P-Value:", p_value_score)
```

```python
alpha = 0.05
if p_value_score < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

```python
plt.figure(figsize=(8,5))
plt.hist(scores, bins=6, color='lightgreen', edgecolor='black')
plt.axvline(mean_score, color='blue', linewidth=2, label='Sample Mean')
plt.axvline(hypothesized_mean_score, color='red', linestyle='--', linewidth=2, label='Hypothesized Mean')
plt.legend()
plt.title("Exam Scores: Sample Mean vs Hypothesized Mean")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.show()
```
